```python
import cv2

# img1 = cv2.imread("Boston.jpeg")
# img2 = cv2.imread("Boston1.jpeg")
img1 = cv2.imread("Boston.jpeg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("Boston1.jpeg", cv2.IMREAD_GRAYSCALE)
```

```python
print(img1 is None, img2 is None)
```

```python
sift = cv2.SIFT_create(nfeatures=100,
nOctaveLayers=4,
contrastThreshold=0.06,
edgeThreshold=15,
sigma=1.9)
keypoints1, descriptors1 = sift.detectAndCompute(img1, None)
keypoints2, descriptors2 = sift.detectAndCompute(img2, None)
```

```python
img1_with_kp = cv2.drawKeypoints(
    img1,               # original image
    keypoints1,         # keypoints detected
    None,               # output image, None to create new
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS  # draws size and orientation
)
img2_with_kp = cv2.drawKeypoints(
    img2,               # original image
    keypoints2,         # keypoints detected
    None,               # output image, None to create new
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS  # draws size and orientation
)
```

```python
cv2.imshow('keypoints1',img1_with_kp)
cv2.imshow('keypoints2',img2_with_kp)

```

```python
# orb = cv2.ORB_create()
# keypoints1, descriptors1 = orb.detectAndCompute(img1, None)
# keypoints2, descriptors2 = orb.detectAndCompute(img2, None)
```

```python
bf = cv2.BFMatcher()
matches = bf.knnMatch(descriptors1, descriptors2, k=2)


good_matches = []
for m, n in matches:
    if m.distance < 0.25 * n.distance:
        good_matches.append(m)
```

```python
matched_img = cv2.drawMatches(
    img1, keypoints1,
    img2, keypoints2,
    good_matches, None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

cv2.imshow('Matches', matched_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
total_matches = len(matches)
good_matches_count = len(good_matches)
bad_matches_count = total_matches - good_matches_count

accuracy = good_matches_count / total_matches * 100

print(f"Total matches: {total_matches}")
print(f"Good matches: {good_matches_count}")
print(f"Bad matches: {bad_matches_count}")
print(f"Accuracy: {accuracy:.2f}%")
```

Camera Calibration: We can undistort and calibrate an image as if it were a camera. This method uses edge detection to unblur an image.
Pose Estimation: We can use edge detection to draw 3D objects in a 2D image.
Epipolar Geometry: 2 images of the same objects taken at different angles will draw epilines connected to epipoles that coincide with the epilines on the other image. 
Depth Map from Stereo Images: Use 2 copies of the image, one is smaller to create a depth estimation heatmap of the image. 
### 2: How depth information can be inferred from feature matches
Feature matching algorithms use epipolar geometry, drawing connections between the same points on different images. The algorithm can also create a depth estimation using one or two different images. As long as the image is calibrated properly, the algorithm is able to detect edges to create estimations.
